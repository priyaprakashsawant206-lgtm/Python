```python
#elif statement
age=int(input("Enter your age:"))
if (age<18):
    print("Minor")
elif (age>18) and (age<60):
    print("Adult")
elif (age>=60) and (age<100):
    print("Senior Citizen")
else:
    print("Invalid")
    
```

```python
#grading system
marks = int(input("Enter your marks: "))

if marks <= 0 or marks > 100:
    print("Invalid Marks")
elif marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B+")
elif marks >= 60:
    print("B")
elif marks >= 50:
    print("C")
elif marks >= 35:
    print("Pass")
else:
    print("Fail")
```

```python
#Rating system
rating=int(input("Enter Restaurant rating (1-5):"))

if (rating==5):
    print("Excellent")
elif(rating==4):
    print("Very Good")
elif(rating== 3):
    print("Good")
elif(rating==2):
    print("Average")
elif(rating==1):
    print("Poor")
else:
    print("Invalid Rating")
```

```python
#nested if else
nationality=input("Enter the nationality:")
age=int(input("Enter the age:"))
if (nationality.lower()=="indian"):   #.lower() converts the input to lowercase before comparing.
    print("Indian Citizen")
    if(age>=18 and age<100):
        print("Eligible to vote")
    else:
        print("Not eligible to vote")
else:
    print("Not a Citizen")
```

```python
#continue ----->continue skips the current iteration and moves to the next one.
i=0
while i<5:
    i=i+1
    if(i==3):
        continue
    print(i)
```

```python
#break ---> break completely stops the loop when a condition is met.
i=0
while i<5:
    i=i+1
    if(i==3):
        break
    print(i)
```

```python
#palindrome
def palindrome():
    word = input("Enter a word: ") 
    if word==word[::-1]:
        print ("Palindrome")
    else:
        print ("Not a Palindrome")
```

```python
palindrome()
```
