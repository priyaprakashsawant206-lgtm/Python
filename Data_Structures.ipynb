```python
# Stores unique values and is useful for removing duplicates or performing set operations.
```

```python
s1.pop()
```

```python
len(s1)
```

```python
s1.clear()
```

```python
s1
```

```python
s1={1,2,3,4,5}
s2={1,2,3,4,7}
```

```python
#INTERSECTION : common elements 
s1.intersection(s2)
```

```python
#UNION:all the elements from both sets
s1.union(s2)
```

```python
#DIFFERENCE: elements in s1 but not s2
s1.difference(s2)
```

```python
#DIFFERENCE: elements in s2 but not s1
s2.difference(s1)
```
