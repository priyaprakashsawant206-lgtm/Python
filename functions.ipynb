```python
# Question 4: Reverse String
# Write a function `reverse_string(s)` that returns the reversed string.
# reverse_string("hello") -> "olleh"
def rev_str():
    name=input("Enter your name:")
    return name[::-1]
rev_str() 
```

```python
# Question 5: Find Maximum
# Write a function `find_max(numbers)` that returns the largest number.
# find_max([3, 7, 2, 9, 4]) -> 9
def find_max(numbers):
    maximum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum

print(find_max([3, 7, 2, 9, 4]))
```

```python
# Question 7: Is Palindrome
# Write a function `is_palindrome(s)` that returns True if the string
# reads the same forwards and backwards.
# is_palindrome("racecar") -> True
def palindrome():
    word = input("Enter a word: ") 
    if word==word[::-1]:
        print ("Palindrome-TRUE")
    else:
        print ("Not a Palindrome-FALSE")
palindrome()
```

```python
# Question 8: Count Vowels
# Write a function `count_vowels(s)` that returns the number of vowels
# (a, e, i, o, u) in a string.
# count_vowels("hello world") -> 3
def count_vowels(s):
    count=0
    for char in s:
        if char in "aeiou":
            count=count+1
    return count
text = input("Enter a string: ")
print("Number of vowels:", count_vowels(text))
    
```

```python
# Question 9: Factorial
# Write a function `factorial(n)` that returns n! (n factorial).
def fact(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact
n = int(input("Enter a number: "))
print("Factorial is:", fact(n))
```

```python
# Question 10: Is Prime
# Write a function `is_prime(n)` that returns True if n is a prime number.
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False
    return True
n = int(input("Enter a number: "))

print(is_prime(n))
    
```

```python
# Question 11: Celsius to Fahrenheit
# Write a function `celsius_to_fahrenheit(c)` that converts Celsius to
# Fahrenheit using the formula: F = C * 9/5 + 32
# celsius_to_fahrenheit(0) -> 32.0
def celsius_to_fahrenheit(c):
    return c * 9/5 + 32
c = float(input("Enter temperature in Celsius: "))

print("Temperature in Fahrenheit:", celsius_to_fahrenheit(c))
```

```python
# Question 12: Count Occurrences
# Write a function `count_occurrences(lst, item)` that returns how many times `item` appears in `lst`.
# count_occurrences([1, 2, 2, 3, 2], 2) -> 3

def count_occurrences(list, item):
    count = 0

    for x in lst:
        if x == item:
            count = count + 1
    return count
list = [1, 2, 2, 3, 2]
item = 2
print(count_occurrences(lst, item))
```

```python
# Question 13: Average of List
# Write a function `average(numbers)` that returns the average of a list.
# average([2, 4, 6]) -> 4.0
def average(numbers):
    total = 0
    for num in numbers:
        total = total + num

    return total / len(numbers)
numbers = [2, 4, 6]
print(average(numbers))
```

```python
# Question 15: FizzBuzz
# Write a function `fizzbuzz(n)` that returns "Fizz" if n is divisible
# by 3, "Buzz" if divisible by 5, "FizzBuzz" if divisible by both,
# otherwise returns str(n).
# fizzbuzz(15) -> "FizzBuzz"
# fizzbuzz(9)  -> "Fizz"
# fizzbuzz(7)  -> "7"
def fizzbuzz(n):
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return str(n)

```

```python
fizzbuzz(9)
```

```python
fizzbuzz(7)
```
