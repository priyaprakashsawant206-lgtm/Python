### if-else Statement

```python
# if else
age=int(input("Enter your age:"))
if age>18 and age<=100:
    print("Adult")
else:
    print("Minor")
```

### if-elif-else statement

```python
#elif statement
age=int(input("Enter your age:"))
if (age<18):
    print("Minor")
elif (age>18) and (age<60):
    print("Adult")
elif (age>=60) and (age<100):
    print("Senior Citizen")
else:
    print("Invalid")
    
```

```python
#grading system
marks = int(input("Enter your marks: "))

if marks <= 0 or marks > 100:
    print("Invalid Marks")
elif marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B+")
elif marks >= 60:
    print("B")
elif marks >= 50:
    print("C")
elif marks >= 35:
    print("Pass")
else:
    print("Fail")
```

```python
#Rating system
rating=int(input("Enter Restaurant rating (1-5):"))

if (rating==5):
    print("Excellent")
elif(rating==4):
    print("Very Good")
elif(rating== 3):
    print("Good")
elif(rating==2):
    print("Average")
elif(rating==1):
    print("Poor")
else:
    print("Invalid Rating")
```

### Nested if statement

```python
#nested if else
nationality=input("Enter the nationality:")
age=int(input("Enter the age:"))
if (nationality.lower()=="indian"):   #.lower() converts the input to lowercase before comparing.
    print("Indian Citizen")
    if(age>=18 and age<100):
        print("Eligible to vote")
    else:
        print("Not eligible to vote")
else:
    print("Not a Citizen")
```

```python
#Bank loan approval → Country → Employment → Salary → Credit score
country=input("Enter your country:")
employment=input("Are you employed: (yes/no):")
salary=int(input("Enter your salary:"))
credit_score=int(input("Enter your credit-score:"))
if (country.lower()=="india"):
    if(employment.lower()=="yes"):
        if(salary>=30000):
            if(credit_score>=700):
                print("Loan Approved")
            else:
                print("Loan Rejected-low credit score")
        else:
            print("Loan Rejected-low salary")
    else:
        print("Loan Rejected-Employment required")
else:
    print("Loan Rejected-only indian citizen are eligible")
            
```
