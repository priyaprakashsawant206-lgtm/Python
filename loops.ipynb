# LOOPS

LOOPS -----> A loop is used to repeat a block of code multiple times.

```python
#while loop
i=0
while i<5:
    print(i)
    i+=1
    
```

```python
i=0
while i<5:
    i+=1
    print(i)
```

```python
#odd numbers
i=1
while i<=15:
    print(i)
    i=i+2
```

```python
#even numbers
i=2
while i<15:
    print(i)
    i=i+2
```

```python
#10 to 1
i=10
while i>1:
    print(i)
    i=i-1
```

```python
#continue -----> continue skips the current iteration and moves to the next one.
i=0
while i<5:
    i=i+1
    if(i==3):
        continue
    print(i)
```

```python
#break ---> break completely stops the loop when a condition is met.
i=0
while i<5:
    i=i+1
    if(i==3):
        break
    print(i)
```

```python
# for loop -----> Used when iterating over a sequence/range
num=[1,3,4,55,66,76,34,21,32,45,90,100]
for i in num:
    print(i)
```

```python
#even numbers from num
for i in num:
    if (i%2==0):
        print(i)
```

```python
#odd numbers from num
for i in num:
    if (i%2!=0):
        print(i)
```

```python
#both odd and even numbers
for i in num:
    if (i%2==0):
        print(i,"is even") 
    else:
        print(i,"is odd")
    
```

```python
#print hello 5 times
i=1
while i<=5:
    print("HELLO")
    i=i+1
```

```python
# range ---->range() is used to generate a sequence of numbers, mainly with for loops.
for i in range(10):
    print(i)
```

```python
#odd numbers
for i in range(1,12,2):
    print(i)
```

```python
#even numbers
for i in range(2,12,2):
    print(i)
```
